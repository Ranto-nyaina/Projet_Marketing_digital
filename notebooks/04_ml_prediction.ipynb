```python
import pandas as pd

# Charger les données
customers = pd.read_csv(
    "../data/raw/customers_data.csv"
)

sales = pd.read_csv(
    "../data/raw/sales_data.csv"
)
```

```python
# Calculer le revenu
sales["Revenue"] = (
    sales["Quantity"] *
    sales["Sale_Price"]
)
```

```python
# Créer les variables clients
features = sales.groupby(
    "Customer_ID"
).agg(
    Number_of_Purchases=("Sale_ID", "count"),
    Total_Quantity=("Quantity", "sum"),
    Total_Revenue=("Revenue", "sum")
).reset_index()

display(features)
```
